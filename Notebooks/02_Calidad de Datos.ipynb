##### Categorización de empleo

1. **Normalizar**: minúsculas, sin puntuación y espacios unificados ("Registered Nurse," = "registered nurse").
2. **Mapear por palabras clave**: las palabras clave salen de los títulos y tokens más frecuentes en train. Se buscan como palabra completa (`\b`) para que "rn" no coincida dentro de "intern" ni "it" dentro de "title".
3. **Resolver ambigüedades por prioridad**: los títulos mezclan sector ("nurse") y nivel jerárquico ("manager"). Criterio: **el sector manda sobre el nivel**. Las categorías se evalúan en el orden del diccionario y gana la primera que coincide, por eso `direccion` va al final ("nurse manager" -> salud, "sales manager" -> ventas) y las profesiones específicas antes que las genéricas ("legal assistant" -> legal, no administrativo).
4. **Medir cobertura**: porcentaje de títulos no nulos que no caen en `otros`.

El mapeo solo usa conocimiento de dominio (no el target), así que no hay leakage. En producción hay que aplicar la misma función a los clientes nuevos.

```python
def normalizar_empleo(serie):
    return (serie.str.lower()
                 .str.replace(r'[^a-z0-9 ]', ' ', regex = True)
                 .str.replace(r'\s+', ' ', regex = True)
                 .str.strip())

empleo_norm = normalizar_empleo(cat['empleo'])
print('Únicos antes:', cat['empleo'].nunique())
print('Únicos tras normalizar:', empleo_norm.nunique())
```

```python
# El orden importa: gana la primera categoría que coincide.
# Profesiones específicas primero, roles genéricos (administrativo) y nivel jerárquico (direccion) al final.
PATRONES_SECTOR = {
    'legal': ['attorney', 'lawyer', 'paralegal', 'legal', 'counsel', 'judge'],
    'salud': ['nurse', 'nursing', 'rn', 'lpn', 'lvn', 'cna', 'physician', 'doctor', 'md',
              'medical', 'pharmacist', 'pharmacy', 'therapist', 'dental', 'dentist',
              'hygienist', 'clinical', 'hospital', 'patient', 'caregiver', 'care giver',
              'health', 'healthcare', 'surgical', 'surgeon', 'radiologic', 'radiology',
              'paramedic', 'emt', 'phlebotomist', 'veterinarian', 'veterinary', 'lab',
              'laboratory', 'technologist', 'optometrist', 'optician', 'chiropractor',
              'pathologist'],
    'educacion': ['teacher', 'professor', 'school', 'instructor', 'education', 'educator',
                  'tutor', 'faculty', 'librarian', 'university', 'college',
                  'paraprofessional', 'trainer'],
    'finanzas': ['accountant', 'accounting', 'bookkeeper', 'financial', 'finance', 'banker',
                 'bank', 'loan', 'mortgage', 'underwriter', 'auditor', 'controller', 'tax',
                 'payroll', 'credit', 'insurance', 'claims', 'billing', 'teller', 'cfo',
                 'cpa', 'investment', 'collector', 'collections', 'accounts payable',
                 'accounts receivable', 'compliance'],
    # 'officer' a secas no se incluye: chocaría con "chief executive officer" o "loan officer"
    'seguridad_publica': ['police', 'sheriff', 'deputy', 'correctional', 'corrections',
                          'correction officer', 'probation', 'firefighter', 'fire',
                          'sergeant', 'lieutenant', 'captain', 'trooper', 'detective',
                          'investigator', 'special agent', 'law enforcement', 'military',
                          'army', 'navy', 'air force', 'marine', 'usps', 'postal',
                          'mail carrier', 'letter carrier', 'city carrier', 'rural carrier',
                          'federal', 'government', 'county', 'city of', 'state of',
                          'security'],
    'tecnologia': ['software', 'developer', 'programmer', 'it', 'network', 'systems',
                   'system', 'data', 'web', 'database', 'engineer', 'engineering',
                   'architect', 'devops', 'computer', 'technology', 'desktop', 'helpdesk',
                   'scientist', 'chemist'],
    'transporte': ['driver', 'truck', 'trucker', 'bus', 'pilot', 'flight', 'dispatcher',
                   'warehouse', 'logistics', 'delivery', 'courier', 'forklift', 'shipping',
                   'freight', 'transportation', 'conductor', 'carrier', 'mail handler',
                   'mailhandler', 'material handler', 'packer'],
    'hosteleria': ['server', 'bartender', 'cook', 'chef', 'food', 'restaurant', 'waiter',
                   'waitress', 'hotel', 'housekeeper', 'housekeeping', 'barista',
                   'kitchen', 'catering'],
    'oficios': ['mechanic', 'electrician', 'welder', 'carpenter', 'machinist', 'operator',
                'laborer', 'labor', 'foreman', 'forman', 'maintenance', 'technician', 'tech',
                'construction', 'plumber', 'installer', 'production', 'plant', 'assembler',
                'assembly', 'hvac', 'painter', 'lineman', 'custodian', 'janitor',
                'fabricator', 'manufacturing', 'mason', 'inspector', 'millwright',
                'pressman', 'contractor', 'quality control', 'quality assurance'],
    'ventas': ['sales', 'salesman', 'retail', 'cashier', 'store', 'realtor', 'real estate',
               'merchandiser', 'broker', 'buyer', 'marketing', 'service advisor'],
    'administrativo': ['assistant', 'secretary', 'receptionist', 'clerk', 'clerical',
                       'administrative', 'admin', 'administrator', 'administration',
                       'office', 'customer service', 'csr', 'coordinator', 'hr',
                       'human resources', 'recruiter', 'processor', 'representative', 'rep',
                       'specialist', 'analyst', 'associate', 'consultant', 'advisor',
                       'planner', 'estimator'],
    'direccion': ['manager', 'mgr', 'gm', 'management', 'director', 'owner', 'president',
                  'ceo', 'coo', 'cto', 'vp', 'executive', 'supervisor', 'partner',
                  'superintendent', 'chief', 'lead', 'leader', 'principal', 'founder',
                  'head'],
}

# Una regex por categoría; \b obliga a coincidir con la palabra completa
REGEX_SECTOR = {sector: r'\b(?:' + '|'.join(palabras) + r')\b'
                for sector, palabras in PATRONES_SECTOR.items()}
```

```python
def coincidencias_sector(empleo_norm):
    """Matriz booleana: qué categorías coinciden con cada título normalizado."""
    return pd.DataFrame({sector: empleo_norm.str.contains(regex, regex = True, na = False)
                         for sector, regex in REGEX_SECTOR.items()})

def asignar_sector(empleo_norm):
    coincidencias = coincidencias_sector(empleo_norm)
    # idxmax devuelve la primera columna True, es decir, la categoría con más prioridad
    sector = coincidencias.idxmax(axis = 1).where(coincidencias.any(axis = 1), 'otros')
    return sector.mask(empleo_norm.isna(), 'desconocido')

coincidencias = coincidencias_sector(empleo_norm)
cat['sector_empleo'] = asignar_sector(empleo_norm)
cat['sector_empleo'].value_counts(normalize = True)
```

```python
# Casos de control: confirmar que la prioridad asigna lo esperado
casos = pd.Series(['sales manager', 'nurse manager', 'it manager', 'office manager',
                   'legal assistant', 'medical assistant', 'loan officer',
                   'chief executive officer', 'police officer', 'teacher'])
pd.DataFrame({'empleo': casos, 'sector': asignar_sector(casos)})
```

```python
no_nulos = empleo_norm.notna().sum()
cobertura = 1 - (cat['sector_empleo'] == 'otros').sum() / no_nulos
ambiguos = coincidencias.sum(axis = 1) > 1

print(f'Cobertura sobre títulos no nulos: {cobertura:.1%}')
print(f'Títulos que coinciden con más de una categoría: {ambiguos.sum() / no_nulos:.1%}')

# Combinaciones de categorías más frecuentes entre los ambiguos: revisar que la prioridad resuelva bien
coincidencias[ambiguos].apply(lambda fila: ' + '.join(fila.index[fila]), axis = 1).value_counts().head(10)
```

```python
# Títulos más frecuentes que quedaron en 'otros': de acá salen las palabras clave de la próxima iteración
empleo_norm[cat['sector_empleo'] == 'otros'].value_counts().head(30)
```

### Atípicos

#### Por desviación típica

##### Identificación

```python
num_desv_tip = 4
```

Función que devuelve los índices de los atípicos

```python
def atipicos_desv_tip(variable, num_desv_tip = 4):
    #sacamos los nulos por ahora
    variable = variable.dropna()
    #calculamos los límites
    media = np.mean(variable)
    sd = np.std(variable)
    umbral = sd * num_desv_tip
    lim_inf = media - umbral
    lim_sup = media + umbral
    #encontramos los índices de los que están fuera de los límites
    indices = [indice for indice,valor in variable.items() if valor < lim_inf or valor > lim_sup]
    return(indices)
```

Función que cuenta el número de atípicos

```python
def conteo_atipicos(df,variable, num_desv_tip = 4):
    atipicos = atipicos_desv_tip(df[variable], num_desv_tip)
    return(df.loc[atipicos,variable].value_counts().sort_index())
```

Obtener los 10 primeros atipicos ordenados en cada variable

```python
for variable in var_atipicos_dt:
    print('\n' + variable + ':\n')
    conteo = conteo_atipicos(num, variable, num_desv_tip)
    print(conteo.sort_index().head(10))
```

#### Por rango de negocio

##### Identificación

Valores que no tienen sentido de negocio:

* dti: ratio de deuda sobre ingresos, esperable entre 0 y 100. El valor 999 es un código de "sin dato", no un ratio
* num_hipotecas: más de 7 es extremadamente raro
* porc_uso_revolving: un porcentaje de uso no puede superar 100

```python
pd.DataFrame({
    'regla': ['dti == 999', 'dti < 0', 'dti > 100 (sin 999)', 'num_hipotecas > 7', 'porc_uso_revolving > 100'],
    'casos': [(num.dti == 999).sum(),
              (num.dti < 0).sum(),
              ((num.dti > 100) & (num.dti != 999)).sum(),
              (num.num_hipotecas > 7).sum(),
              (num.porc_uso_revolving > 100).sum()]
})
```

Conclusiones:

* dti == 999: es un código, no un valor --> eliminar los registros (son muy pocos)
* resto de valores fuera de rango --> recortar a los límites de negocio

Si no se corrige, el MinMaxScaler de la fase de transformación toma 999 como máximo y comprime al 99% de los clientes en una franja muy pequeña de la escala.

##### Corrección

```python
# eliminar los registros con dti 999 (código de sin dato), tanto de cat como de num
a_eliminar = num.loc[num.dti == 999].index.values
cat = cat[~cat.index.isin(a_eliminar)]
num = num[~num.index.isin(a_eliminar)]

# recortar a los límites de negocio
num['dti'] = num['dti'].clip(0, 100)
num['num_hipotecas'] = num['num_hipotecas'].clip(0, 7)
num['porc_uso_revolving'] = num['porc_uso_revolving'].clip(0, 100)

num[['dti', 'num_hipotecas', 'porc_uso_revolving']].describe().T
```

## PRÉSTAMOS SIN DESENLACE

La target del modelo PD se construye a partir de `estado`. Solo tiene sentido para préstamos que ya terminaron: o se pagaron (Fully Paid) o se perdieron (Charged Off / Default).

Los préstamos vigentes (Current), en periodo de gracia o con atraso todavía no tienen desenlace. Si se codifican como 0 ("buen pagador") se asume algo que no ocurrió y la tasa de impago queda subestimada.

### Identificación

```python
cat.estado.value_counts()
```

### Corrección

```python
estados_sin_desenlace = ['Current', 'In Grace Period', 'Late (16-30 days)', 'Late (31-120 days)']

a_eliminar = cat.loc[cat.estado.isin(estados_sin_desenlace)].index.values
cat = cat[~cat.index.isin(a_eliminar)]
num = num[~num.index.isin(a_eliminar)]

print('Registros eliminados:', len(a_eliminar))
print('Registros restantes:', len(cat))
cat.estado.value_counts()
```
